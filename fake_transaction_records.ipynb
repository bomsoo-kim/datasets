```python
import random

def generate_fake_date(
    types = ['TYPE1','TYPE2','TYPE3','TYPE4','TYPE5','TYPE6','TYPE7','TYPE8'],
#     years = list(range(1880,2021)),
    years = list(range(1900,2021)), #update: 2021-05-27
    months = list(range(1,13)),
    months_char = ['JAN', 'FEB', 'MAR', 'APR', 'MAY', 'JUN', 'JUL', 'AUG', 'SEP', 'OCT', 'NOV', 'DEC', 'JANUARY', 'FEBRUARY', 'MARCH', 'APRIL', 'JUNE', 'JULY', 'AUGUST', 'SEPTEMBER', 'OCTOBER', 'NOVEMBER', 'DECEMBER'],
    dates = list(range(1,32)),
):
    # random date generation
    date_type = random.choice(types)
    year = str(random.choice(years))
    month = str(random.choice(months))
    month_char = random.choice(months_char)
    date = str(random.choice(dates))
    sep1 = random.choice([',', ' ', '/', '.', '-'])
    padding = random.choice([1,2])

    if date_type == 'TYPE1': # ex) '06/02/1986', '2/07/1986', '12/02/1967',
        random_date = month.zfill(padding) + sep1 + date.zfill(padding) + sep1 + year
    elif date_type == 'TYPE2': # ex) '1986/06/02'
        random_date = year + sep1 + month.zfill(padding) + sep1 + date.zfill(padding)
    elif date_type == 'TYPE3': # ex) '19860602'
        random_date = year + month.zfill(2) + date.zfill(2)
    elif date_type == 'TYPE4': # ex) '06021986'
        random_date = month.zfill(2) + date.zfill(2) + year
    elif date_type == 'TYPE5': # ex) '05 MAR 1980'
        random_date = date.zfill(padding) + sep1 + month_char.zfill(padding) + sep1 + year
    elif date_type == 'TYPE6': # ex) 'MAR 05, 1980'
        random_date = month_char.zfill(padding) + sep1 + date.zfill(padding) + sep1 + year
    elif date_type == 'TYPE7': # ex) '1980 MAR 05'
        random_date = year + sep1 + month_char.zfill(padding) + sep1 + date.zfill(padding)
    elif date_type == 'TYPE8': # 'ex) 02-SEP-98'
        random_date = date.zfill(padding) + sep1 + month_char.zfill(padding) + sep1 + year[-2:]

    # random DOB prefix
    sep2 = random.choice([',', ' ', '/', '.', '-',':'])
    sep3 = random.choice(['',' '])
    dob_prefix = random.choice(['DOB','DATE OF BIRTH','BIRTHDAY','BIRTH DATE','BIRTH']) + sep2 + sep3

    return date_type, random_date, dob_prefix

if __name__=='__main__':
    random.seed(123)
    for _ in range(5):
        date_type, random_date, dob_prefix = generate_fake_date()
        print(date_type, ' ::: ', random.choice(['',dob_prefix]) + random_date)
```
